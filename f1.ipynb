I want to get the all season sessions. loop through each race, get results of each race.  

```python
import fastf1 
import pandas as pd 

winners = [] #Storing (year, round, event, winner)

for year in [2022,2023,2024]:
    calendar = fastf1.get_event_schedule(year)

    for _, row in calendar.iterrows(): #Loop to iterate rows of the data
        rnd = row["RoundNumber"]
        event = row["EventName"]

        if "Testing" in str(event):
            continue

        try: 
            session = fastf1.get_session(year, int(rnd), "R")
            session.load(
                laps = False
                telemetry= False
                weather= False
                messages= False
            )

            results = session.results

            if results is None or results.empty: 
                continue

            winner = results.loc[0,"FullName"]
            winners.append((year,int(rnd), event, winner))
        
        except Exception as e:
            #If a race cant load then skip and continue
            continue 

df_winners = pd.DataFrame(winners,columns=["Year", "Round", "Event", "Winner"])
df_winners.head()
 


```
